# 專案週期 3：雙樣本推論分析

### 研究問題 5：性別與身高
### 目標：利用 2007 年青少年風險行為調查 (YRBS) 資料集，確定男女學生的平均身高是否有顯著差異。

## 階段一：數據預處理與資料清理 (Data Preprocessing & Cleaning)
### 專案主題：青少年性別與身高之顯著差異分析 (YRBS 2007)
本筆記本將針對 2007 年美國 YRBS 原始數據進行篩選、去噪與標準化，為後續的統計分析奠定乾淨的數據基礎。

### 本階段處理步驟
1. 讀取原始數據 `YRBS_2007.csv`。
2. 篩選 `WhatIsYourSex` (性別) 與 `HowTallAreYouWithoutShoesInMeters` (身高)。
3. 處理缺失值 (NaN) 與生理學上不合理的異常身高值（小於 0.5m 或大於 2.5m）。
4. 根據 Codebook 將性別重新編碼：`1` -> `Female`, `2` -> `Male`。
5. 將乾淨數據儲存至 `data/processed/cleaned_height_data.csv`。

```python
import pandas as pd
import numpy as np
import os

print("啟動階段一：資料清理與預處理...")

# 自動偵測路徑：避免 Windows 權限問題，優先在當前目錄尋找與建立
raw_filename = 'YRBS_2007.csv'
possible_paths = [f'data/raw/{raw_filename}', raw_filename]
raw_path = next((p for p in possible_paths if os.path.exists(p)), None)

if raw_path:
    df = pd.read_csv(raw_path)
    print("成功讀取原始資料")
    print(f"原始資料筆數: {len(df)} 筆")
else:
    raise FileNotFoundError("找不到原始 CSV 檔案，請確認 YRBS_2007.csv 已放在與本程式同目錄，或 data/raw/ 中。")
```

```python
# 1. 選取欄位並移除缺失值
target_cols = ['WhatIsYourSex', 'HowTallAreYouWithoutShoesInMeters']
clean_df = df[target_cols].dropna().copy()
clean_df.columns = ['Gender', 'Height']

# 2. 移除不合理身高 (過濾 Outliers)
clean_df = clean_df[(clean_df['Height'] > 0.5) & (clean_df['Height'] < 2.5)]

# 3. 性別重新編碼 (1=Female, 2=Male)
clean_df['Gender_Label'] = clean_df['Gender'].map({1: 'Female', 2: 'Male'})

print(f"清理完成！有效分析樣本數: {len(clean_df)} 筆")
print(f"移除了 {len(df) - len(clean_df)} 筆缺失/異常樣本")
```

```python
# 自動建立當前目錄下的資料夾 
output_dir = 'data/processed'
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(output_dir, 'cleaned_height_data.csv')
clean_df.to_csv(output_path, index=False)

print(f"乾淨數據已成功存檔至: {output_path}")
print("階段一完成！請下載此 CSV 檔案")
```
