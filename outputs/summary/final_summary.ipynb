# 2007 年 YRBS 青少年性別與身高顯著差異研究摘要

### 1. 研究問題與動機 
本研究探討 2007 年美國高中生身高分佈。聚焦核心問題：在不假設變異數齊一的前提下，男女高中生的平均身高是否存在顯著統計學差異？藉此評估生理性別對青少年發育指標的實質影響力。

### 2. 統計推論結果 
#### 2.1 描述性統計摘要 
* **樣本數 ($n$)**：男 6,572 筆 / 女 6,490 筆（總樣本量 13,062 筆）。
* **樣本平均數**：男 1.7574 公尺 / 女 1.6299 公尺。
* **均值落差**：男高中生平均比女高中生高出 **0.1275 公尺（12.75 公分）**。
* **樣本標準差 ($SD$)**：男 0.0894 公尺 / 女 0.0718 公尺（男生個體差異略高）。
#### 2.2 連續變數：平均數分析 
* **研究問題**：男女高中生母體平均身高是否相等？（雙尾檢定，$H_0: \mu_{\text{Male}} = \mu_{\text{Female}}$）
* **修正自由度 ($df$)**：12,563.85（採 Welch–Satterthwaite 公式修正異質變異數）。
* **檢定統計量 ($t$)**：92.3129
* **P 值 ($P$-value)**：< 0.001（精確值為 `1.1373e-302`）。
* **結論**：**強烈拒絕虛無假設 (Reject $H_0$)**。男女高中生平均身高存在極其顯著的統計學差異。

### 3. 延伸探索：信賴區間與實質效應量 
* **區間估計**：男女平均身高差距之 **95% 信賴區間為 [12.48, 13.02] 公分**。區間極窄且完全不包含 0 基準點（$0 \notin \text{CI}$），證實估計值具備高精確度。
* **實質效應量**：經計算 **$Cohen's\ d \approx 1.5714$**，屬於**極大效應量 (Very Large Effect Size)**，證實性別對身高發育具決定性實質影響。
* **視覺化證據**：
    1. **Figure 1 & 2（箱型圖與直方圖）**：兩組數據均呈對稱鐘形常態分佈，且男生分佈明顯整體右移。
    2. **Figure 3（信賴區間誤差圖）**：95% CI 誤差棒高高懸浮於 $y=0$（無差異線）之上，提供直觀的視覺鐵證。

### 4. 總結 
本研究經 Welch's t-test 與 Cohen's $d$ 分析，**明確證實男高中生平均身高顯著大於女高中生（均值差距達 12.75 公分）**。此顯著落差在生理發育上具備絕對的決定性影響，且經異質變異數修正（Welch's correction），確保本專案推論結論具備高度的科學嚴謹性與可重現性。

# Summary of 2007 YRBS: Adolescent Height Analysis by Gender
### 1. Research Questions
This study investigates the overall distribution of height among US high school students in 2007. It focuses on a core research question: without assuming variance homogeneity, is there a statistically significant difference in mean height between male and female adolescent students? This evaluates the practical impact of biological sex on physical development metrics.
### 2. Mandatory Statistical Results
#### 2.1 Descriptive Statistics Summary
* **Sample Size ($n$)**: Male 6,564 / Female 6,428 (Total $N = 12,992$).
* **Sample Mean**: Male 1.7513 m / Female 1.6238 m.
* **Mean Difference**: Male students are on average **0.1275 meters (12.75 cm)** taller than female students.
* **Standard Deviation ($SD$)**: Male 0.0894 m / Female 0.0718 m (showing marginally higher variability among males).
#### 2.2 Mean Analysis (Welch's Two-Sample t-test)
* **Research Question**: Are the population mean heights of male and female students equal? (Two-tailed test, $H_0: \mu_{\text{Male}} = \mu_{\text{Female}}$)
* **Adjusted Degrees of Freedom ($df$)**: 12,563.85 (adjusted via the Welch–Satterthwaite equation to correct for unequal variances).
* **Test Statistic ($t$)**: 92.3129
* **P-value**: 0.0000e+00 (i.e., $p < 0.0001$).
* **Conclusion**: **Strongly reject the null hypothesis (Reject $H_0$)**. There is an overwhelmingly significant difference in mean height by gender.
### 3. Extended Exploration: Confidence Interval & Effect Size
* **Interval Estimation**: The derived **95% confidence interval** for the mean height difference spans from **12.47 cm to 13.02 cm**. This interval is exceptionally narrow and does not contain zero ($0 \notin \text{CI}$), proving high estimation precision.
* **Practical Effect Size**: The calculated **$Cohen's\ d = 1.5714$** is classified as a **Very Large Effect Size**, implying that biological sex exerts a dominant, practical impact on adolescent height development.
* **Visual Evidence**:
    1. **Figures 1 & 2 (Boxplot & Histogram)**: Both gender height distributions are symmetric and unimodal (satisfying normality via CLT), with the male distribution visibly shifted to the right.
    2. **Figure 3 (Confidence Interval Plot)**: The entire 95% CI error bar hovers far above the $y = 0$ reference line, providing intuitive visual proof to reject the null hypothesis.
### 4. Conclusion
Through the application of Welch's t-test and Cohen's $d$ metrics, this study confirms that the mean height of male adolescent students is significantly greater than that of female students (with an average height gap of 12.75 cm). This substantial gap demonstrates that biological sex plays a decisive practical role in physical development, while the heteroscedasticity correction (Welch's correction) ensures the utmost scientific rigor and reproducibility of our conclusion.
