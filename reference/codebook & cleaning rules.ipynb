# 重新編碼規則與清洗規則 
本檔案紀錄將原始 `YRBS_2007.csv` 資料庫中的原始代號，進行重新編碼與數據清洗之流程。

## 1. 原始變數編碼對照表
| 原始資料庫欄位 | 原始問卷定義與邏輯描述 | 專案清洗後欄位名 | 轉換後資料型態 |
| :--- | :--- | :--- | :--- |
| **`q2`** | What is your sex?（性別代號） | `Gender_Label` | 類別變數 (String) |
| **`q3`** | How tall are you without shoes in meters?（身高） | `Mean Height` | 連續變數 (Float) |
### 1.1 類別變數重新編碼邏輯 (`q2`)
* 原始問卷代碼 `1`  重新編碼為字串 `Female`
* 原始問卷代碼 `2`  重新編碼為字串 `Male`
* 空白 / 遺漏值  統一標記為 `NA`（排除分析）
### 1.2 數值變數清洗邏輯 (`q3`)
* 正常的連續身高數值（公尺）一律保留
* 原始問卷中的無效填報代碼 `9.99`  判定為缺失值並予以剔除

## 2. 資料清洗流程與最終有效樣本量
為避免髒資料或無效代碼干擾兩樣本檢定結果，執行以下清洗動作：
1. **完全刪除法**：自動檢查每筆資料，只要 `q2`（性別）或 `q3`（身高）任一欄位出現空白（NaN）或無效代碼（9.99），該樣本一律完全剔除。
2. **清洗後有效分析樣本量**：
   * **有效總樣本量**：**13,062 筆**
   * **Female (女性群組)**：6,490 筆
   * **Male (男性群組)**：6,572 筆

# Codebook & Data Cleaning Rules

This document records the data cleaning process and the re-coding workflow applied to the raw `YRBS_2007.csv` dataset.

## 1. Raw Variable Codebook Table
| Raw Variable Name | Original Survey Coding & Logic | Processed Variable Name | Target Data Type |
| :--- | :--- | :--- | :--- |
| **`q2`** | What is your sex? | `Gender_Label` | Categorical (String) |
| **`q3`** | How tall are you without shoes in meters? | `Mean Height` | Quantitative (Float) |

### 1.1 Categorical Re-coding Logic (`q2`)
* Raw survey code `1` ➡️ Re-coded as string `Female`
* Raw survey code `2` ➡️ Re-coded as string `Male`
* Blanks / Missing values ➡️ Labeled as `NA` (Excluded from analysis)

### 1.2 Quantitative Cleaning Logic (`q3`)
* Valid continuous height entries in meters are retained.
* Raw invalid code `9.99` ➡️ Identified as missing value and filtered out.

## 2. Data Cleaning Workflow & Final Analytic Sample Size
Cleaning rules applied to prevent dirty data or invalid survey codes from distorting the results:
1. **Listwise Deletion**: Observations with missing (`NaN`) or invalid values (`9.99`) in either `q2` or `q3` are completely removed from the analytic sample.
2. **Final Sample Sizes**:
   * **Total Valid Sample Size**: **13,062 rows**
   * **Female Group ($n_1$)**: 6,490 rows
   * **Male Group ($n_2$)**: 6,572 rows
