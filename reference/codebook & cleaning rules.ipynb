# 重新編碼規則 
本檔案紀錄本專案如何將原始資料庫中的代號，進行重新編碼與數據清洗，以符合統計推論之作業基準。
## 1. 正式生理分組變數
* **原始變數**：`q2` (生理性別)
* **本項目使用的編碼**：
  * 原始碼 `1` 重新編碼為 `Gender_Label = Female`
  * 原始碼 `2` 重新編碼為 `Gender_Label = Male`
  * 其他遺漏或空白 標記為 `NA` (剔除)
## 2. 形式連續變數
* **原始變數**：`q3` (身高)
* **本項目使用的清潔劑與過濾邏輯**：
  * 轉換為浮點數數值型（Float）
  * 剔除原始資料中的無效值（如身高填寫為 9.99）
  * 最終用於平均值分析的有效樣本量：13,062 筆 (女性 6,490 筆、男性 6,572 筆)

# Codebook & Cleaning Rules
This codebook documents the data cleaning process and the re-coding workflow applied to the raw CDC YRBSS dataset for inferential statistics.
## 1. Primary Grouping Variable (Biological Sex)
* **Original Variable**: `q2`
* **Re-coding Rules**:
  * Raw code `1` Re-coded as `Gender_Label = Female`
  * Raw code `2` Re-coded as `Gender_Label = Male`
  * Missing values / Blank Labeled as `NA` (Excluded from analysis)

## 2. Continuous Outcome Variable (Height)
* **Original Variable**: `q3`
* **Data Cleaning & Filtering Workflow**:
  * Converted data entries into floating-point numbers (`Float`).
  * Filtered out invalid survey codes (such as height filled as `9.99`).
  * **Final Sample Size**: Female = 6,490 rows; Male = 6,572 rows; Total Valid Sample = 13,062 rows.
