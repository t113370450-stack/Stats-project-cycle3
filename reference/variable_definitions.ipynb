# 變數定義 
本專案針對 2007 年 YRBS 數據中的生理性別與身高發育狀況進行分析，以下為核心變數之定義：
| 原始變數名稱 | 變數描述 | 測量方式 / 單位 | 分析樣本數 (n) |
| :--- | :--- | :--- | :--- |
| **`q2`** | 受訪高中生的生理性別 | 類別變數：1 (女性)、2 (男性) | 13,062 (排除缺失值後) |
| **`q3`** | 受訪高中生的身高 | 連續變數：公尺 (m) | 13,062 (排除缺失值後) |
* 數據品質說明：所有分析均已排除原始數據中的缺失值（NaN），確保統計之準確性。

# Variable Definitions
This project analyzes the relationship between biological sex and height distribution using the 2007 YRBSS dataset. The definitions of the core variables are as follows:
| Original Variable | Variable Description | Measurement / Unit | Analytic Sample Size (n) |
| :--- | :--- | :--- | :--- |
| **`q2`** | Respondent's biological sex | Categorical: 1 (Female), 2 (Male) | 13,062 (after listwise deletion) |
| **`q3`** | Respondent's height | Continuous: Meters (m) | 13,062 (after listwise deletion) |
* **Data Quality Note**: All analyses have excluded missing values (NaN) from the original data to ensure the accuracy and statistical power of the test.
