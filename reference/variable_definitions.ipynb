# 變數定義與統計方法依據 
本文件紀錄 Project Cycle 3（第五題：生理性別與身高關係）之變數定義、統計方法選擇原因，以及統計假設之檢驗。


## 1. 核心變數定義
### 1.1 分組變數 / 獨立變數
* **原始欄位名稱**：`q2`（重新編碼為 `Gender_Label`）
* **變數描述**：受訪高中生的生理性別
* **資料型態**：類別變數 / 二元變數
* **群組定義**：區分為兩個獨立群體：女性群組（$n_1 = 6,490$）與男性群組（$n_2 = 6,572$）。
### 1.2 依變數 / 反應變數
* **原始欄位名稱**：`q3`（重新編碼為 `Mean Height`）
* **變數描述**：受訪高中生不穿鞋的實際身高
* **資料型態**：連續變數 / 數值變數
* **計量單位**：公尺 (m)

## 2. 統計方法選擇
* **選用方法**：**Welch's Two-Sample t-test（異質變異數兩樣本 t 檢定）**
* **選擇依據**：
  反應變數（身高）屬於數值型變數。根據指導原則，在比較兩組平均數時，除非有明確證據支持兩組母體變異數相等，否則預設優先使用 Welch's t-test。由於男女高中生的身高樣本變異數並不相等，採用 Welch's t-test 進行自由度修正，以確保統計推論結論的可靠性。

## 3. 統計假設檢驗
在進行推論統計前，針對兩樣本 t 檢定之各項統計假設評估如下：
1. **獨立性**：數據源自 CDC YRBSS 的隨機抽樣調查，且男女兩個性別群體相互排斥，滿足獨立樣本假設。
2. **測量尺度**：反應變數（身高）為連續數值變數，符合 t 檢定的資料型態要求。
3. **常態性 / 樣本量**：樣本量極大（女性 6,490 筆、男性 6,572 筆），根據中央極限定理 (CLT)，樣本平均數之抽樣分佈趨近於常態分佈，滿足 T 檢定之常態性要求。
4. **變異數同質性**：不假設男女兩組的母體變異數相等。直接採用 Welch's 異質變異數修正公式調整自由度，以防範變異數不相等帶來的統計偏誤。

# Variable Definitions & Method Justification

This document details the variable definitions, selection of statistical methods, and assumption considerations for Project Cycle 3 (Question 5: Gender and Height).

## 1. Core Variable Definitions
### 1.1 Group / Independent Variable
* **Original Variable Name**: `q2` (Recoded as `Gender_Label`)
* **Variable Description**: Respondent's biological sex.
* **Data Type**: Categorical / Binary.
* **Group Definition**: Divided into two independent populations: Female ($n_1 = 6,490$) and Male ($n_2 = 6,572$).

### 1.2 Response / Dependent Variable
* **Original Variable Name**: `q3` (Recoded as `Mean Height`)
* **Variable Description**: Respondent's height without shoes.
* **Data Type**: Continuous / Quantitative.
* **Unit of Measurement**: Meters (m).

## 2. Statistical Method Choice
* **Selected Method**: **Welch's Two-Sample t-test**
* **Justification**: 
  The response variable (height) is quantitative. According to the guidelines for Project Cycle 3, Welch's t-test must be used by default when comparing two means unless there is a clear reason to assume equal variances. Since the sample variances of height between male and female students are unequal, Welch's t-test is applied to handle variance heterogeneity.

## 3. Assumptions Considered
Key assumptions required for a two-sample t-test are evaluated as follows:
1. **Independence**: Data is obtained from the CDC YRBSS random survey; the two gender groups are mutually exclusive and independent.
2. **Measurement Scale**: The response variable (height) is a continuous quantitative variable, matching the data requirement for a t-test.
3. **Normality / Sample Size**: The sample sizes are extremely large ($n_1 = 6,490$ for females, $n_2 = 6,572$ for males). According to the Central Limit Theorem (CLT), the sampling distribution of the sample means approximates a normal distribution.
4. **Homogeneity of Variance**: Equal variance is NOT assumed. Welch's adjustment for degrees of freedom is applied to prevent statistical bias.
